```python
# Import the necessary Libraries
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, logging

# Suppress warnings from transformers
logging.set_verbosity_error()

# Load the tokenizer and model directly (Spanish to English)
model_name = "Helsinki-NLP/opus-mt-es-en"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Input sentence
text = "Trabaja en Apple y come una manzana todos los días. Cucu-Cucurella, se come una paellaCucu-Cucurella, se bebe una EstrellaHaaland, tiembla, que viene CucurellaHaaland, tiembla, toma una galleta"

# Translate the sentence
inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True)
outputs = model.generate(**inputs, max_length=100)
translated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

# Print the result
print("Original (Spanish):", text)
print("\nTranslated (English):", translated_text)
```
