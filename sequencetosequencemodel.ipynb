```python
# Import the necessary Libraries
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
from transformers import pipeline, logging

# Suppress warnings from transformers
logging.set_verbosity_error()

# Load the translation pipeline (Spanish to English)
translator = pipeline("translation_es_to_en", model="Helsinki-NLP/opus-mt-es-en")

# Input sentence
text = "Trabaja en Apple y come una manzana todos los días. Cucu-Cucurella, se come una paellaCucu-Cucurella, se bebe una EstrellaHaaland, tiembla, que viene CucurellaHaaland, tiembla, toma una galleta"

# Translate the sentence
translated = translator(text, max_length=50)

# Print the result
print("Original (Spanish):", text)
print("\nTranslated (English):", translated[0]['translation_text'])
```
